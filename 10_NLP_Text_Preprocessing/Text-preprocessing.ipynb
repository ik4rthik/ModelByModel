```python
import nltk
import string
import re
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

# Download NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')

nlp = spacy.load("en_core_web_sm")
```

```python
text = """
My name is Karthik S. I am from Kochi, Kerala.
I completed my schooling at Central Calvathy School, Fort Kochi.
I belong to a family of four members.
Currently, I am pursuing my final year B.Tech in Artificial Intelligence at Adi Shankara Institute of Engineering and Technology.
I have completed internships at InfineCode Tech and Genzee Technology in Robotics.
I also participated in a hackathon conducted at SCMS.
My hobbies are travelling, listening to music and watching movies.
I am interested in Artificial Intelligence, Machine Learning and software development.
My strengths are quick learning, problem-solving skills and adaptability.
One of my weaknesses is stage fear, but I am working to improve my public speaking skills.
"""
```

```python
lower_text = text.lower()

print(lower_text)
```

```python
no_punctuation = lower_text.translate(str.maketrans('', '', string.punctuation))

print(no_punctuation)
```

```python
no_numbers = re.sub(r'\d+', '', no_punctuation)

print(no_numbers)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
tokens = word_tokenize(no_numbers)

print(tokens)
```

```python
stop_words = set(stopwords.words("english"))

filtered_words = [word for word in tokens if word not in stop_words]

print(filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print(lemmatized_words)
```

```python
import nltk

nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_tags = pos_tag(filtered_words)

print(pos_tags)
```

```python
doc = nlp(text)

print("Named Entities:\n")

for entity in doc.ents:
    print(entity.text, "---->", entity.label_)
```

```python
final_text = " ".join(lemmatized_words)

print(final_text)
```

```python
print("Original Length :", len(text.split()))
print("After Stopword Removal :", len(filtered_words))
print("Final Preprocessed Length :", len(final_text.split()))
```
